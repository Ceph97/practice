### Tuples
- Immutable and hashables sequence of objects

- a = ```"a", "b", "c"```
- You dont have to put paranthesis, but for readability enclose them with paranthesis ```()```
- One element tuple should end with a ```comma ,``` without comma its going to be a string.


````python
newTuple = ("a", "b", "c", "d", "e")
print(newTuple)
````

````python
# Tuple() constructor treats each char as an element
newTuple = tuple("abcde")
print(newTuple)
````

````python
# Accessing Tuple
newTuple = tuple("abcde")
newTuple[1]
````

````python
# Accessing Tuple
newTuple = tuple("abcde")
newTuple[1:]
````

````python
# Traversing a Tuple
# Accessing Tuple
newTuple = tuple("abcde")

for i in newTuple:
    print(i)

# Range(len(tuple)) works too 
````

````python
# Searching an element in Tuple

# in operator
newTuple = tuple("abcde")
# Time O(n)
print("a" in newTuple)

# .index()
print(newTuple.index("a"))


# custom
def searchTuple(p_tuple, element):
    for i in range(0, len(p_tuple)):
        if p_tuple[i] == element:
            return i, element
    return -1, -1
````

````python
# Tuple functions
newTuple = tuple("abcde")
newTuple2 = tuple("uvwxyz")

# + will add elements together
print(f"+ operator: {newTuple + newTuple2}")

# * will mirror the tuple and combines elements
print(f"* operator: {newTuple * 2}")

# Methods that add or remove elements are not available as its immutable

# count()
# The occurrency of an element
print(f".count(): {newTuple.count('a')}")

# count()
# Returns index of element
print(f".index(): {newTuple.index('e')}")

````
