# SHORTEN URL
- Given long Url: https://google.com/long/url Output: https://short.ly/<code>
- Given short url: https://short.ly/<code> Output: https://google.com/long/url

```python
import random
import string

class ShortenUrl:
    def __init__(self):
        self.long_url_store = {}
        self.short_url_store = {}
        self.randomSource = string.ascii_letters + string.digits#"jsduvifvy1673t782387"
        self.urlString = "https://shot.rl/"

    def generateShortUrl(self,longUrl:str)->str:
        
        if longUrl in self.long_url_store:
            return f"{self.urlString}{self.long_url_store.get(longUrl)}"
        
        # Then we need generate and store the code. Stringfy it also
        # Choices does return a []
        randomString = ''.join(random.choices(self.randomSource, k=6))
        # In rare case where random matches in store, prevent collisions
        while randomString in self.short_url_store:
            # regenarate a new random string
            randomString = ''.join(random.choices(self.randomSource, k=6))
       
        # save it
        self.short_url_store[randomString] = longUrl
        self.long_url_store[longUrl] = randomString

        return f"{self.urlString}{randomString}"
    
    def getLongUrl(self, shorUrl:str)->str:
        urlcode = shorUrl.split('/')
        # Doing a check to avoid unsupported urls
        if not shortUrl.startswith(self.urlString):
            return f"Invalid URL string, it should be {self.urlString}"
        return self.short_url_store.get(urlcode[-1],"long url not found")

shortenUrl = ShortenUrl()
longUrl = "http://google.com/this/is/very/long/url"
shortUrl = shortenUrl.generateShortUrl(longUrl)
print(f'Short url is {shortUrl}')
print(f'Long url is {shortenUrl.getLongUrl(shortUrl)}')


```
