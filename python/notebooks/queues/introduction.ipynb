## Queues

- A DS that works on data in a first in first out manner. FIFO


- Can implement it with ```List``` if size is defined it is more efficient as no assigning new memory at runtime or ```LinkedList```
- Operations:
    - Create Queue
    - Enqueue: Add element at the back
    - Dequeue: Remove element in front
    - Peek: Returns first element without removing
    - isEmpty
    - isFull
    - deleteQueue

````python
# Using List without Size limit

class Queue:
    def __init__(self):
        self.items = []
    
    def __str__(self):
        values = [f"[{x}]" for x in self.items]
        return '<-'.join(values)
    
    def __len__(self):
        return len(self.items)
    
    def isEmpty(self):
        if self.items == []:
            return True
        return False

    # Time Armotised O(1) as growing queue affect time
    # Space O(1)
    def enqueue(self, value):
        self.items.append(value)
        return True
    
    # dequeue() remove at the begining of the queue
    # Time O(n) element shift to left if index 0 is removed
    # Space O(1)
    def dequeue(self):
        if self.isEmpty():
            return False
        return self.items.pop(0)

    # peek() returns first item before
    # Without dequeing it
    def peek(self):
        if self.isEmpty():
            return False
        return self.items[0]
    
    # delete(self):
    def delete(self):
        self.items = None

customQueue = Queue()
customQueue.enqueue(10)
customQueue.enqueue(20)
customQueue.enqueue(30)
customQueue.enqueue(40)
customQueue.enqueue(50)
print(customQueue)
customQueue.dequeue()
customQueue.dequeue()
print(customQueue)
print(customQueue.peek())
print(customQueue)
````

````python
# Circular Queue
# More efficient queueue as there is no item deletion to shift elements
# We put None placeholders.

class Queue:
    def __init__(self, maxSize):
        self.items = [None] * maxSize
        self.maxSize = maxSize
        # Track the index
        # As we do not actually delete items
        # We just update
        # Start of the queue
        self.start = -1
        # End of queue
        self.top = -1
    # Time O(1)
    def __str__(self):
        values = [f"[{x}]" for x in self.items]
        return '<-'.join(values)
    
    # Time and Space O(1)
    def isFull(self):
        if self.top + 1 == self.start:
            return True
        elif self.start == 0 and self.top + 1 == self.maxSize:
            return True
        else:
            return False
    
    # isEmpty()
    # Space and Time O(1)
    def isEmpty(self):
        if self.top == -1:
            return True
        return False
    
    # Time: O(1)
    # Space: O(1)
    def enqueue(self, value):
        if self.isFull():
            return False
        else:
            self.top += 1
            if self.start == -1:
                self.start = 0
        self.items[self.top] = value
        return True
    
    def dequeue(self):
        if self.isEmpty():
            return False
        else:
            firstElement = self.items[self.start]
            start = self.start
            # We will know we have emptied up everything
            if self.start == self.top:
                self.start = -1
                self.top = -1
            # Circling back to the first element space 
            # Since the last space has been reached
            elif self.start +1  == self.maxSize:
                self.start = 0
            else:
                self.start += 1
            self.items[start] = None
            return firstElement
    
    # Time O(1) Space O(1)
    def peek(self):
        if self.isEmpty():
            return False
        else:
            return self.items[self.start]
    
    def delete(self):
        self.items = self.maxSize * [None]
        self.top = -1
        self.start = -1


customQueue = Queue(3)
print(customQueue.isEmpty())
customQueue.enqueue(10)
customQueue.enqueue(20)
customQueue.enqueue(30)
print(customQueue)
customQueue.dequeue()
print(customQueue)
print(customQueue.peek())
customQueue.delete()
print(customQueue)

````

````python
# Linked List Implementation:
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None
    
    def __str__(self):
        return str(self.value)

class LinkedList:
    def __init__(self):
        self.head = None
        self.tail = None
    
    # Makes the list iterable
    def __iter__(self):
        curNode = self.head
        while curNode:
            # it returns the current iter without exiting out
            # Makes it usable in a for loop else where
            yield curNode
            curNode = curNode.next

class Queue:
    def __init__(self):
        # Time and Space O(1)
        self.linkedList = LinkedList()
    
    def __str__(self):
            values = [f"[{x}]" for x in self.linkedList]
            return '<-'.join(values)
    
    def isEmpty(self):
        if self.linkedList.head == None:
            return True
        return False
        
    
    # Time O(1)
    # Space O(1)
    def enqueue(self, value):
        new_node = Node(value)
        if self.linkedList.head is None:
            self.linkedList.head = new_node
            self.linkedList.tail = new_node
        else:
            # Current tail is no longer tail
            # Since last node.next is None, make it point to new Node
            # Making new node the absolute last node
            self.linkedList.tail.next = new_node
            # Then point tail to new_node as its the new tail
            self.linkedList.tail = new_node
    
    # dequeue()
    # Time O(1)
    # Space O(1)
    def dequeue(self):
        if self.isEmpty():
            return False
        else:
            node_to_delete = self.linkedList.head
            # Check if only one element in the queue
            # If Head and Tail points to the same node
            # That means after dequeue we need to reset the queue to empty
            if self.linkedList.head == self.linkedList.tail:
                self.linkedList.head = None
                self.linkedList.tail = None
            else:
                # Point the head to the second node
                self.linkedList.head = self.linkedList.head.next
                node_to_delete.next = None
        # Return the deleted node
        return node_to_delete
    
    # peek() return values without removing it from the queue
    def peek(self):
        if self.isEmpty():
            return False
        return self.linkedList.head
    
    # delete()
    # Space and Time O(1)
    def delete(self):
        self.linkedList.head = None
        self.linkedList.tail = None


custQueue = Queue()
custQueue.enqueue(10)
custQueue.enqueue(20)
custQueue.enqueue(30)
custQueue.enqueue(40)
print(custQueue)
custQueue.dequeue()
print(custQueue.dequeue())
print(custQueue)
print(custQueue.peek())


````

````python
# Using deque class
# You can use it for stack or queue

from collections import deque

# FIFO queue

custQ = deque(maxlen=3)

custQ.append(1)
custQ.append(2)
custQ.append(3)
print(custQ)
# Dequeue
print(custQ.popleft())

# clear
print(custQ.clear())

print(custQ)
````

````python
# Using Queue module
# Multi producer multi consumer
#  We will use LIFO queue, LIFO queue is also available for stacks
import queue as q

custQue = q.Queue(maxsize=3)

# append
custQue.put(1)
custQue.put(2)
custQue.put(3)

# check its full
print(custQue.full())
# Dequeue
print(custQue.get())
# size
print(custQue.qsize())
````

````python
# Using multiprocessing module
# Allows for a shared queue between processes etc
# Same methods as standalone queue module

from multiprocessing import Queue as q

custQue = q(maxsize=3)

# append
custQue.put(1)
custQue.put(2)
custQue.put(3)

# check its full
print(custQue.full())
# Dequeue
print(custQue.get())
# size
# custQue.qsize()
````
