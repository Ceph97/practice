## Linked list (Singly linked list)

Linked List is a form of a sequential collection and it does not have to be in order. A Linked

list is made up of independent nodes that may contain any type of data and each node has a

reference to the next node in the link.

```python
class Node:
    def __init__(self, value):
        # Stores the actual value.
        self.value = value
        # Reference to next pointer
        # Set to None as we do not know the next neighbour.
        self.next = None


class LinkedList:
    def __init__(self):
        # Initially tail and head are the same node
        # new_node = Node(value)
        # self.head = new_node
        # self.tail = new_node
        # self.length = 1

        # Empty linked list
        self.head = None
        self.tail = None
        self.length = 0
    
    # String representation of the object
    # The Print() on the object will call this
    def __str__(self):
        return self.traverse()
    
    def __len__(self):
        return self.length

    # Add at the end of the linked list
    # TIME: O(1) and SPACE O(1)
    def append(self, value):
        new_node = Node(value)
        if self.head is None:
            self.head = new_node
            self.tail = new_node
        else:
            self.tail.next = new_node
            self.tail = new_node

        # increment counter
        self.length += 1
        return True
    
    # Add a node at the begining of list
    # new_node → old head → rest of list
    # TIME: O(1) AND SPACE: O(1)
    def prepend(self, value):
        new_node = Node(value)
        if self.head is None:
            self.head = new_node
            self.tail = new_node
        else:
            new_node.next = self.head
            self.head = new_node
        self.length += 1
        return True
    
    # Add a node at any given index
    #  TIME: O(n) SPACE: O(1)
    def insert(self, index, value):
        new_node = Node(value)
        # It should stay within bounds
        if index < 0 or index > self.length:
            return False
        # edge case for 0 index or length is 0
        if index == 0 or self.length == 0:
            # Since logic is already implemented, we can use prepend
            self.prepend(value)

        # The Main insert logic
        else:
            temp_node = self.head
            # Before the current index
            # If you want new node at 2
            # Stop before the 2
            for _ in range(index-1):
                # Update the next node reference until you reach location you want to insert
                temp_node = temp_node.next
            new_node.next = temp_node.next
            temp_node.next = new_node
            self.length += 1
        return True
    
    # Visits all nodes in the linked list
    # Time O(n) Space O(1)
    def traverse(self) -> str:
        result = ''
        current = self.head
        while current is not None:
            result += str(current.value)
            if current.next is not None:
                result += "->"
            # increment to next pointer until it hits a None
            current = current.next
        return result
    
    # Search a node in a linked list
    # Its Binary search algorithm
    # Time O(n) and Space is O(1)
    def search(self, value):
        current = self.head
        index = 0
        while current is not None:
            if current.value == value:
                return index
            current = current.next
            index += 1
        return -1
    
    # Get method, we pass in the index and get values
    # Return the value else returns -1
    # TIME O(n), SPACE O(1)
    def get(self, index):
        current = self.head
        counter = 0
        #  -1 should return last element which is the tail
        if index == -1:
            return self.tail
        if index < -1 or index >= self.length:
            return None

        while current is not None:
            if counter == index:
                return current
            current = current.next
            counter += 1
        return None
    
    # Updates the specified index with the passed value
    # Time O(n), Space O(1) 
    def set(self, index:int, value):
        current = self.head
        counter = 0
        #  -1 should return last element which is the tail
        if index == -1:
            if self.tail is not None:
                self.tail.value = value
                return True
        if index < -1 or index >= self.length:
            return False

        while current is not None:
            if counter == index:
                current.value = value
            current = current.next
            counter += 1
        return True
    
    # popfirst(), removes first element and returns the value
    # Time O(1), Space O(1)
    def popfirst(self):
        if self.head is None:
            return None
        temp = self.head.value
        self.head = self.head.next
        self.length -= 1
        # If list becomes empty reset tail also
        if self.length == 0:
            self.tail = None
        return temp
    
    # pop(), removes the last node of the list which is essentially the tail
    # Traverse to the node before the last node the break the the link to tail
    # Time O(n) and Space O(1) 
    def pop(self):
        if self.head is None:
            return None
        
        last_node = self.tail
        temp = self.head

        # Only one node is present
        if self.head == self.tail:
            self.head = None
            self.tail = None
            self.length -= 1
            return last_node.value

        while temp is not None:
            if temp.next == last_node:
                temp.next = None
                self.tail = temp
                self.length -= 1
                break

            temp = temp.next

        return last_node.value
    
    # remove(), specific element at specific index
    # Node before should skip the reference of the node we want to remove
    # TIME O(n) SPACE O(1)
    def remove(self, index: int):
        if index == 0:
            return self.popfirst()
        if index >= self.length or index < 0:
            return None
        if index == self.length -1:
            return self.pop()
        # a node before which is current index minus 1
        prev_node = self.get(index-1)
        node_to_delete = prev_node.next

        # Break the chain
        prev_node.next = node_to_delete.next
        node_to_delete.next = None 

        # reduce the length
        self.length -= 1
        return node_to_delete
    
    # Deleteall()
    # Set tail and head to None
    # Garbage collector will purge everything
    # TIME O(1), SPACE O(1)
    def deleteall(self):
        self.head = None
        self.tail = None
        self.length = 0




new_linked_list = LinkedList()
new_linked_list.append(10)
new_linked_list.append(20)
new_linked_list.append(50)
new_linked_list.append({"hello":1})
print(new_linked_list)
new_linked_list.prepend("Head")
print(new_linked_list)
new_linked_list.insert(5,"head")
print(new_linked_list)
new_linked_list.append({"hello":1})
print(new_linked_list)
print("============Traverse()==========")
print(new_linked_list.traverse())
print("============Search(value)==========")
print(f"search value 50 index is: {new_linked_list.search(50)}")
print("============get(index)==========")
print(f"get() index of 1: {new_linked_list.get(-1)}")
print("============set(index,value)==========")
new_linked_list.set(0,"First")
print(new_linked_list)
print("============popfirst()==========")
print(new_linked_list.popfirst())
print(new_linked_list)
print(new_linked_list.popfirst())
print(new_linked_list)
print("============pop()==========")
print(new_linked_list.pop())
print(new_linked_list)
print("============remove()==========")
new_linked_list.remove(0)
print(new_linked_list)


```
