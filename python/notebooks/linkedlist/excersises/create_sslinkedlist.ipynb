## Singly Linked List

Write code in the **language of your choice** to implement a **singly linked list**.

A singly linked list should have the following properties:

- Each **node** contains a piece of data.
- The **`Node`** class constructor takes a **`value`** as an argument and initializes the node's **`value`** attribute.
- Each node holds a **reference (link)** to the **next node** in the list.
- The **`next`** attribute should initially be set to **`None`**. It will store a reference to the next node in the list.
- The **`LinkedList`** class constructor takes a **`value`** as an argument.
- The constructor creates a new **`Node`** object using the provided **`value`**.
- The **`head`** and **`tail`** attributes should both point to the newly created node.
- The **`length`** attribute should be initialized to **`1`**, representing the current number of nodes in the linked list.

```python
class Node:
    def __init__(self, value):
        # Actual value
        self.value = value
        # Reference to next node
        self.next = None

class LinkedList:
    def __init__(self, value):
        new_node = Node(value)
        self.head = new_node
        self.tail=new_node
        self.length = 1
```
