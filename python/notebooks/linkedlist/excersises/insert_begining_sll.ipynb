## Insertion at the Beginning of a Singly Linked List

Write a function to insert a new element at the beginning of a singly linked list. 

LinkedList and Node class are already provided.

Note: Function name must be prepend

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None
        self.length = 0
    
    def __str__(self):
        temp_node = self.head
        result = ''
        while temp_node is not None:
            result += str(temp_node.value)
            if temp_node.next is not None:
                result += ' -> '
            temp_node = temp_node.next
        return result
        
    # Implement Here
    def prepend(self, value):
        # Create the new node first
        new_node = Node(value)
        # Check if its an empty list
        if self.head is None:
            self.head = new_node
            self.tail = new_node
        # Else
        else:
            new_node.next = self.head
            self.head = new_node

        self.length += 1
    
    # Add at the end of the linked list
    def append(self,value):
        new_node = Node(value)
        # Check if its an empty list
        if self.head is None:
            self.head = new_node
            self.tail = new_node
        else:
            self.tail.next = new_node
            self.tail = new_node
        self.length += 1
    
    def remove(self, index):
        if index < 0 or index >= self.length:
            return None
           # if node to be removed is the head node
        elif index == 0:
            popped_node = self.head
            if self.length == 1:
                self.head = None
                self.tail = None
            else:
                self.head = self.head.next
            popped_node.next = None
            self.length -= 1
            return popped_node
        else:
            # Traverse the list upto the index
            # Change the previous node to skip the node at index
            # Use the next reference of the node we want to delete
            # This will leave current index node orphaned hence deletion
            current = self.head
            # We need to traverse the list until
            # We are a node behind what we want to delete
            for _ in range(index - 1):
                current = current.next

            # node_to_purge is pointing to the node we want to delete  
            node_to_purge = current.next
            # this mean we are setting a node behind to point to node after
            # node_to_purge is orphan
            current.next = node_to_purge.next
            node_to_purge.next = None
        
            self.length -= 1
            return node_to_purge
    
    def reverse(self):
        previous = None
        current = self.head
        
        # The tail will become the head
        self.tail = self.head

        while current:
            next_node = current.next
            current.next = previous
            
            # For the next iteration
            previous = current
            current = next_node
        
        # Whatever the last node is it will become the head
        self.head = previous

        
        



sll = LinkedList()
sll.prepend(20)
sll.append(50)
sll.append(60)
sll.append(70)
sll.append(80)
print(sll)
sll.remove(2)
print(sll.reverse())
print(sll)
```
