## Linked list

Linked List is a form of a sequential collection and it does not have to be in order. A Linked

list is made up of independent nodes that may contain any type of data and each node has a

reference to the next node in the link.

```python
class Node:
    def __init__(self, value):
        # Stores the actual value.
        self.value = value
        # Reference to next pointer
        # Set to None as we do not know the next neighbour.
        self.next = None


class LinkedList:
    def __init__(self):
        # Initially tail and head are the same node
        # new_node = Node(value)
        # self.head = new_node
        # self.tail = new_node
        # self.length = 1

        # Empty linked list
        self.head = None
        self.tail = None
        self.length = 0
    
    # String representation of the object
    # The Print() on the object will call this
    def __str__(self):
        temp_node = self.head
        result = ''
        while temp_node is not None:
            result += str(temp_node.value)
            if temp_node.next is not None:
                result += '->'
            temp_node = temp_node.next
        return result
    
    def __len__(self):
        return self.length

    # Add at the end of the linked list
    # TIME: O(1) and SPACE O(1)
    def append(self, value):
        new_node = Node(value)
        if self.head is None:
            self.head = new_node
            self.tail = new_node
        else:
            self.tail.next = new_node
            self.tail = new_node

        # increment counter
        self.length += 1
        return True
    
    # Add a node at the begining of list
    # new_node → old head → rest of list
    # TIME: O(1) AND SPACE: O(1)
    def prepend(self, value):
        new_node = Node(value)
        if self.head is None:
            self.head = new_node
            self.tail = new_node
        else:
            new_node.next = self.head
            self.head = new_node
        self.length += 1
        return True
    
    # Add a node at any given index
    #  TIME: O(n) SPACE: O(1)
    def insert(self, index, value):
        new_node = Node(value)
        # It should stay within bounds
        if index < 0 or index > self.length:
            return False
        # edge case for 0 index or length is 0
        if index == 0 or self.length == 0:
            # Since logic is already implemented, we can use prepend
            self.prepend(value)
        else:
            temp_node = self.head
            # Before the current index
            # If you want new node at 2
            # Stop before the 2
            for _ in range(index-1):
                # Update the next node reference until you reach location you want to insert
                temp_node = temp_node.next
            new_node.next = temp_node.next
            temp_node.next = new_node
            self.length += 1
        return True



        

    
        


new_linked_list = LinkedList()
new_linked_list.append(10)
new_linked_list.append(20)
new_linked_list.append(50)
new_linked_list.append({"hello":1})
print(new_linked_list)
new_linked_list.prepend("Head")
print(new_linked_list)
new_linked_list.insert(5,"head")
print(new_linked_list)
new_linked_list.append({"hello":1})
print(new_linked_list)
```
