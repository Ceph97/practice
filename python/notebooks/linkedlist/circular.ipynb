## Circular Linked List

- Same a singly linked list however the last node will have a pointer to the first element in the list essentially creating a circle.

- Its a never ending loop

```python
class Node:
    def __init__(self, value):
        # Stores the actual value.
        self.value = value
        # Reference to next pointer
        # Set to None as we do not know the next neighbour.
        self.next = None


class CSLinkedList:
    # Time and space is O(1) for empty initialization
    def __init__(self):
        # new_node = Node(value)
        # new_node.next = new_node
        # self.head = new_node
        # self.tail = new_node
        # self.length = 1

        self.head = None
        self.tail = None
        self.length = 0

    def __str__(self):
        return self.traverse()
    
    def __len__(self) -> int:
        return self.length

    def traverse(self) -> str:
        result = ''
        current = self.head
        while current is not None:
            result += str(current.value)
            current = current.next
            if current.next == self.head:
                break 
            result += " -> "
            # increment to next pointer until it hits a None
            
        return result
    
    # append(value) add value at the end of the linked list
    def append(self, value):
        new_node = Node(value)
        # if its empty
        if self.length == 0:
            self.head = new_node
            self.tail = new_node
            # it should reference itself to make it circular
            new_node.next = new_node
        else:
            self.tail.next = new_node
            # make it circular
            new_node.next = self.head
            # Make the tail point to the new node
            self.tail = new_node

        self.length += 1
    
    # prepend(value) add value at the begining of linked list
    def prepend(self, value):
        new_node = Node(value)
        if self.head is None:
            self.head = new_node
            self.tail = new_node
            # it should reference itself to make it circular
            new_node.next = new_node
        else:
            new_node.next = self.head
            self.head = new_node
            # Make it circular
            self.tail.next = new_node
        self.length += 1
        return True
    
    # Add a node at any given index
    #  TIME: O(n) SPACE: O(1)
    def insert(self, index, value):
        new_node = Node(value)
        # It should stay within bounds
        if index < 0 or index > self.length:
            return False
        # edge case for 0 index or length is 0
        if index == 0 or self.length == 0:
            # Since logic is already implemented, we can use prepend
            self.prepend(value)

        # The Main insert logic
        else:
            temp_node = self.head
            # Before the current index
            # If you want new node at 2
            # Stop before the 2
            for _ in range(index-1):
                # Update the next node reference until you reach location you want to insert
                temp_node = temp_node.next
            new_node.next = temp_node.next
            temp_node.next = new_node
            self.length += 1
        return True




cslinkedlist = CSLinkedList()
cslinkedlist.append(10)
cslinkedlist.append(20)
cslinkedlist.append(30)
cslinkedlist.append(40)
cslinkedlist.append(50)
# print(cslinkedlist)
cslinkedlist.prepend(1)
cslinkedlist.prepend(3)
cslinkedlist.prepend(3)
cslinkedlist.prepend(3)
cslinkedlist.prepend(3)
cslinkedlist.prepend(0)
print(cslinkedlist)
len(cslinkedlist)
```
