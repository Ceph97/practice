# 2 SUM
You are given an array of integers nums and an integer target, return indices of the two numbers such that they add up to target.

You may assume that each input would have exactly one solution, and you may not use the same element twice.

You can return the answer in any order.

 

### Example 1:
Input: nums = ```[2,7,11,15]```, target = 9
Output: ```[0,1]```
Explanation: Because nums[0] + nums[1] == 9, we return [0, 1].

### Example 2:
Input: nums = ```[3,2,4]```, target = 6
Output: ```[1,2]```

### Example 3:
Input: nums = ```[3,3]```, target = 6
Output: ```[0,1]```
 

Constraints:

2 <= nums.length <= 104
-109 <= nums[i] <= 109
-109 <= target <= 109
Only one valid answer exists.
 

Follow-up: Can you come up with an algorithm that is less than O(n2) time complexity?

````python
# Not the best approach as it is not linear time complex
def two_sum(nums:list, target:int):
    output = []
    # Start from the start of the list
    # This is an On^2
    for n in range(len(nums)):
        # Start 1 position ahead of the outer loop
        for m in range(n + 1, len(nums)):
            if nums[m] + nums[n] == target:
                output.append(n)
                output.append(m)
                return output
    return output


nums = [2,7,11,15]
target = 9
print(two_sum(nums,target))
````

````python
# Using hashmap for linear search
def two_sum(nums:list, target:int) -> list:
    seen = {}
    
    # STEP 1: Make the num key value using enumerate
    for i, num in enumerate(nums):
        # STEP 2:
        # we want to find the number
        # Easiest way is to find what we need
        # By subtracting current num with targer = remainder we need to solve equation
        needed = target - num
        # STEP 4: 
        # Check if the needed number is already present
        # If thats the case then we have solved it:
        if needed in seen:
            # return the indexes:
            return [seen[needed], i]
        # STEP 3:
        # Save the current number as key in seen dict
        # We use it to check if needed has already been seen
        # Basically flipping the index to value
        seen[num] = i
    

nums = [2,4,11,15]
target = 6
print(two_sum(nums,target))
````

````python
# Without using enumaration in O(n) but with 2 passes in space complexity
def two_sum(nums:list, target:int) -> list:
    seen = {}
    nums_dict = {}
    # Populate to map
    for i in range(len(nums)):
        nums_dict[i] = nums[i]
    
    # Check the map
    for key, value in nums_dict.items():
        needed = target - value
        
        if needed in seen:
            return [seen[needed], key]
        
        seen[value] = key
       

nums = [2,4,11,15]
target = 6
print(two_sum(nums,target))
````
