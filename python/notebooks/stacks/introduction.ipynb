## Stack

- Pile of objects stacked on top of each other
- Last in will be the first out.

- We can implement it using array or linked list

```python
# Using a LIST
    # It more efficient to consider last element as top of stack
    # It is efficient to add or remove end of list as no item shifting

class Stack:
    def __init__(self):
        self.items = []

    def __len__(self):
        return len(self.items)

    # O(n) time
    # O(n) space
    def __str__(self):
        if self.is_empty():
            return None
        values =  [str(x) for x in reversed(self.items)]
        return '\n'.join(values)

    
    def is_empty(self) -> bool:
        # len(self) class since we have a __len__ implementation
        return len(self) == 0
    
    # Push() operation
    # Add elements to the stack
    # It more efficient to consider last element as top of stack
    # It is efficient to add or remove end of list as no item shifting
    # Amorties O(1)
    # SPACE: O(1)
    def push(self, element):
        self.items.append(element)

    # pop()
    # Remove the last element in the stack
    # First in First Out
    # SPACE: O(1) and TIME O(1)
    def pop(self):
        if self.is_empty():
            return None
        return self.items.pop()

    # peek()
    # Returns the last element without removing it from the stack
    # O(1) both space and time
    def peek(self):
        if self.is_empty():
            return None
        return self.items[-1]
    
    # clear()
    # Removes all elements in a stack
    def clear(self):
        self.items = []


# TIME O(1)
my_stack = Stack()
my_stack.push(10)
my_stack.push(20)
my_stack.push(30)
# O(1
print(my_stack)
# pop
print("=========pop()=============")
my_stack.pop()
print(my_stack)
print("=========peek()=============")
print(my_stack.peek())

```

```python
# USING LINKED LIST
# It is more efficient to add an item at the begining of linked list
# Prefered approach is performing operation at the head of the
# Head = Top
# Tail = Bottom, however we dont need a bottom as all operations are done at the head

class Node:
    def __init__(self, value):
        self.value = value
        self.next = None

# LinkedList will be Named stack
class Stack:
    # Time and Space O(1)
    def __init__(self):
        self.top = None
        # For Stack, tail reference is not required
        # self.bottom = None
        self.length = 0
    
    # Implement it at the head
    # push()
    def push(self, value):
        new_node = Node(value)
        # New node next reference should point to the current head
        # Since it will be the first inline after push
        new_node.next = self.top
        # Then the Top/Head reference should point to the new_node
        # Making it the new Top/Head
        self.top = new_node
        self.length += 1

    # pop() removes last element we added
    # Time O(1) Space O(1)
    def pop(self):
        if self.top is None:
            return None
        popped_node = self.top
        # The Top/head node should be the 2nd one which is the top.next
        self.top = self.top.next
        # Then break the chain for popped node
        popped_node.next = None
        self.length -= 1
        return popped_node

    # peek() Returns the top of the stack without removing it.
    # O(1) Time and Space
    def peek(self):
        return self.top
    
    def is_empty(self):
        return self.length == 0

    def clear(self):
        self.top = None
        self.length = 0


new_stack = Stack()
new_stack.push(10)
new_stack.push(20)
new_stack.push(30)
new_stack.push(40)
print(new_stack.pop().value)
        
```
