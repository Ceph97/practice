## Stack Min

- How would you design a stack which, in addition to push and pop, has a function min which

returns the minimum element? Push, pop and min should all operate in O(1).

```python
# Implement using linked list
# Track min/max value class wide For each pop() and push()
# Compute the min and max value on push and have a snapshot of those values
# At node leve

class Node:
    def __init__(self, value, min_value, max_value):
        self.value = value
        self.next = None
        # Track the min and max at that point
        # This avoid issues when you pop values and no values
        #  greater than max, it does not decrement it
        # So just check the head max
        self.min_value = min_value
        self.max_value = max_value

# LinkedList will be Named stack
class Stack:
    # Time and Space O(1)
    def __init__(self):
        self.top = None
        # For Stack, tail reference is not required
        # self.bottom = None
        self.length = 0
        
        
        self.min_value = float("inf")
        self.max_value = float("-inf")
    
    # This should only be calculated during push
    # Since we track the currecnt max at node level
    # On push we update those values directly
    def evaluate_min_max(self, value):
        self.min_value = min(self.min_value, value)
        self.max_value = max(self.max_value, value)
        return self.min_value, self.max_value

    def min(self):
        return self.top.min_value
    
    def max(self):
        return self.top.max_value

    # Implement it at the head
    # push()
    def push(self, value):
        min_v, max_v = self.evaluate_min_max(value)
        new_node = Node(value, min_v, max_v)
        # New node next reference should point to the current head
        # Since it will be the first inline after push
        new_node.next = self.top
        # Then the Top/Head reference should point to the new_node
        # Making it the new Top/Head
        self.top = new_node
        self.length += 1
        

    # pop() removes last element we added
    # Time O(1) Space O(1)
    def pop(self):
        if self.top is None:
            return None
        popped_node = self.top
        # The Top/head node should be the 2nd one which is the top.next
        self.top = self.top.next

        # Update the current min and max value to the current top min, max
        if self.top is None:
            self.min_value = float("inf")
            self.max_value = float("-inf")
        else:
            self.min_value = self.top.min_value
            self.max_value = self.top.max_value
        # Then break the chain for popped node
        popped_node.next = None
        self.length -= 1
        
        return popped_node

    # peek() Returns the top of the stack without removing it.
    # O(1) Time and Space
    def peek(self):
        return self.top
    
    def is_empty(self):
        return self.length == 0

    def clear(self):
        self.top = None
        self.length = 0


new_stack = Stack()
new_stack.push(10)
new_stack.push(20)
new_stack.push(30)
new_stack.push(40)
print(new_stack.max())
new_stack.pop()
print(new_stack.max())
print(new_stack.min())
new_stack.pop()
print(new_stack.max())
        


        
```
