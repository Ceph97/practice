## Longest Substring Without Repeating Characters

### Difficulty: Medium

Given a string ``s``, find the length of the longest substring without duplicate characters.

 

Example 1:

Input: s = ```"abcabcbb"```
Output: 3
Explanation: The answer is ```"abc"```, with the length of ```3```. Note that ```"bca"``` and ```"cab"``` are also correct answers.
Example 2:

Input: s = ```"bbbbb"```
Output: 1
Explanation: The answer is ```"b"```, with the length of ```1```.
Example 3:

Input: s = ```"pwwkew"```
Output: 3
Explanation: The answer is ```"wke"```, with the length of 3.
Notice that the answer must be a substring, ```"pwke"``` is a subsequence and not a substring.
 

Constraints:

0 <= s.length <= 105
s consists of English letters, digits, symbols and spaces.

````python
# Using slidding window algorithm
def lengthOfLongestSubstring(string:str) -> int:
    char_index = {}
    # left window and right window will be the loop iteration
    left = 0

    max_len = 0 #return value
    for right, char in enumerate(string):
        # If the char is already present in the map
        # and the current char index >= left value then we need to shrink
        # left window by one position.
        if char in char_index and char_index[char] >= left:
            left = char_index[char] + 1 #Shrink window from left
        char_index[char] = right
        max_len = max(max_len, right - left + 1)
    return max_len

string="pwwkewabc"  
print(lengthOfLongestSubstring(string=string))
````

Because array indices are 0-based, so the length of a window isn't just right - left, it's off by one.

Simple example:

string = ```"pwwkew"```
index =   0 1 2 3 4 5

When ```left=2, right=4``` (window is ```"wke"```):

```right - left``` = ```4 - 2``` = 2 ← wrong, that's only 2 chars
```right - left + 1``` = ```4 - 2 + 1``` = 3 ← correct, "wke" is 3 chars

Think of it this way — if you're standing on index 2 and walking to index 4, you step on 3 tiles (2, 3, 4), not 2. The +1 counts the tile you're already standing on.
