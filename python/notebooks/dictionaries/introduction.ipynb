```python
# Time: O(N)
# Spane:O(N)
# Using dict constructor
eng_span = dict(
    one='uno',
    two='dos',
    three='tres',
)

print(eng_span)
```

```python
# Time: O(N)
# Spane:O(N)

# Using curly bracket
eng_span = {'one': 'uno', 'two': 'dos', 'three': 'tres'}
print(eng_span)
```

```python
# Time: O(N)
# Spane:O(N)

# Using a list of tuple
eng_span_list = [('one','uno'), ('two','dos'),( 'three','tres')]
print(dict(eng_span_list))
```

```python
# Update or add a new element in a dictionary

# Space: O(1)
# Time: O(1)

myDict  = { "name": "Edy", "age":"26"}
myDict['age'] = 27
print(myDict)
```

```python
# Trarverse a dictionary

myDict  = { "name": "Edy", "age":"26", "address":"Dublin"}

# Time: O(n)
# Space: O(1)
def traverseDict(dict:dict):
    for key in dict:
        print(key, dict[key])

traverseDict(myDict)
```

```python
# Find an element in a dict
myDict  = { "name": "Edy", "age":26, "address":"Dublin"}

# Linear search
# Space: O(1)
# Time: O(1)
def searchDict(dict:dict, value):
    for key in dict:
        if dict[key] == value:
            return key, value
    return f"{value}: Non existent"

print(searchDict(myDict, 26))
```

```python
# Delete element from a dict 

myDict  = { "name": "Edy", "age":26, "address":"Dublin"}

# Time: O(1)
# Space: O(1)
del myDict["address"]
print(myDict)
print("==============pop()============")
# You can use pop()
# None is default if key not found
# Delete key-value pair but returns value
# TIME: O(1)
# SPACE: O(1)
remove_element = myDict.pop('name',None)
print(remove_element)
print(myDict)

print("=========popitem()===============")
# You can use popitem()
# It removes the last inserted key-value pair
# TIME AND SPACE O(1)
remove_element = myDict.popitem()
print(remove_element)
print(myDict)

print("=========clear()===============")
# clear()
# Remove all elements from a dictionary
myDict.clear()

```

```python
# Dictionary methods
myDict  = { "name": "Edy", "age":26, "address":"Dublin", "gender":"M"}


# copy()
# Shallow copy of dict and creates a separate dict in memory
myDict2 = myDict.copy()

# fromkeys()
# Return new dict from a provided squence of keys
print("=======fromkeys()===========")
myDict3 = {}.fromkeys([1,2,3],0)
print(myDict3)

# get()
# Returns value of specified key, and return it,
# You can pass a default value if item does not exist, else returns None
print("=======get()===========")
print(myDict3.get(5))

# items()
print("=======items()===========")
# Returns key-value tuple pairs [(key,value),(key2,value2)]
print(myDict.items())

# keys()
print("=======keys()===========")
# Retuens keys of items
print(myDict.keys())

# keys()
print("=======values()===========")
# Retuens values of items
print(myDict.values())


print("==============pop()============")
# You can use pop()
# None is default if key not found
# Delete key-value pair but returns value
# TIME: O(1)
# SPACE: O(1)
remove_element = myDict.pop('gender',None)
print(remove_element)
print(myDict)

print("=========popitem()===============")
# You can use popitem()
# It removes the last inserted key-value pair
# TIME AND SPACE O(1)
remove_element = myDict.popitem()
print(remove_element)
print(myDict)

print("=========clear()===============")
# clear()
# Remove all elements from a dictionary
# myDict.clear()
print("==========setdefault()==========")
# Creates a new key-value pair with the supplied default value if the key is not present
# If key is present it returns its value
print(myDict.setdefault("name","added"))

print("==========update()==========")
# Updates a dict with dict and add new pairs for non existent keys
new_dict = {"height":150, "Shirt Size":"L"}
myDict.update(new_dict)
print(myDict)
```

```python
# Dictionary operations using builting operators
my_dict = {
    0: "three",
    5: "five",
    9: "nine",
    2: "two",
    1 :"one",
    4: ""
}


# in operator
print("============in operator===========")
# default checks keys
print(5 in my_dict)
# to check values
print("nine" in my_dict.values())

print("============len()===========")
# Returns count of the pairs
len(my_dict)

print("============all()===========")
# Returns if all key are true
# 0, "", False are all falsy values
print(all(my_dict.values()))


print("===========sorted()=========")
# sorted(dict)
#  Returns list of keys in a sorted manner
sorted(my_dict)
```
