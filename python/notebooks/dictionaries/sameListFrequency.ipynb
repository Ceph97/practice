
# Same Frequency

Define a function which takes two lists as parameters and check if two given lists have the same frequency of elements.

Example:

list1 = ```[1, 2, 3, 2, 1]```

list2 = ```[3, 1, 2, 1, 3]```

check_same_frequency(list1, list2)

Output:

```False```

````python
def check_same_frequency(list1:list,list2:list) -> bool:
    # if each element in list 1 and list2 appears the exact same amount of time
    def frequency(inputList):
        result = {}
        # calculate the frequency using a hashmap
        for i in inputList:
            # Increment the result
            result[i] = result.get(i,0) + 1
        return result
    
    return frequency(list1) == frequency(list2)

list1 = [1, 2, 3, 2, 1]
list2 = [3, 1, 2, 1, 3]
check_same_frequency(list1, list2)
````
