## Binary Tree


- Binary trees are the data structures in which each node has at most two children,
often referred to as the left and right children.

- All other tree types a derived from binary trees


**WHY**

- Binary trees are a prerequisite for mode advanced trees like BST, AVL, Red Black Trees

- Huffman coding problem, heap priority problem and expression parsing problems can be
solved efficiently using binary trees,


**TYPES OF BINARY TREES**

- Full
- Perfect
- Complete
- Balanced

**OPERATIONS ON A TREE**

- Creation of Tree
- Insertion of a node
- Deletion of a node
- Search for a value
- Traverse all nodes
- Deletion of tree

**TRAVERSING A TREE**

- Depth first search
    - Preorder traversal
    - Inorder traversal
    - Post order traversal
- Breadth first search
    - Level order traversal

```python
# We using the linked list queue to tackle level order traversal

# Linked List Implementation:
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None
    
    def __str__(self):
        return str(self.value)

class LinkedList:
    def __init__(self):
        self.head = None
        self.tail = None
    
    # Makes the list iterable
    def __iter__(self):
        curNode = self.head
        while curNode:
            # it returns the current iter without exiting out
            # Makes it usable in a for loop else where
            yield curNode
            curNode = curNode.next

class Queue:
    def __init__(self):
        # Time and Space O(1)
        self.linkedList = LinkedList()
    
    def __str__(self):
            values = [f"[{x}]" for x in self.linkedList]
            return '<-'.join(values)
    
    def isEmpty(self):
        if self.linkedList.head == None:
            return True
        return False
        
    # Time O(1)
    # Space O(1)
    def enqueue(self, value):
        new_node = Node(value)
        if self.linkedList.head is None:
            self.linkedList.head = new_node
            self.linkedList.tail = new_node
        else:
            # Current tail is no longer tail
            # Since last node.next is None, make it point to new Node
            # Making new node the absolute last node
            self.linkedList.tail.next = new_node
            # Then point tail to new_node as its the new tail
            self.linkedList.tail = new_node
    
    # dequeue()
    # Time O(1)
    # Space O(1)
    def dequeue(self):
        if self.isEmpty():
            return False
        else:
            node_to_delete = self.linkedList.head
            # Check if only one element in the queue
            # If Head and Tail points to the same node
            # That means after dequeue we need to reset the queue to empty
            if self.linkedList.head == self.linkedList.tail:
                self.linkedList.head = None
                self.linkedList.tail = None
            else:
                # Point the head to the second node
                self.linkedList.head = self.linkedList.head.next
                node_to_delete.next = None
        # Return the deleted node
        return node_to_delete
    
    # peek() return values without removing it from the queue
    def peek(self):
        if self.isEmpty():
            return False
        return self.linkedList.head
    
    # delete()
    # Space and Time O(1)
    def delete(self):
        self.linkedList.head = None
        self.linkedList.tail = None
```
