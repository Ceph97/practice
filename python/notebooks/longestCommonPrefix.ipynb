### Longest Common Prefix

Write a function to find the longest common prefix string amongst an array of strings.

If there is no common prefix, return an empty string "".

 

Example 1:

Input: strs = ```["flower","flow","flight"]```
Output: ```"fl"```

Example 2:

Input: strs = ```["dog","racecar","car"]```
Output: ""
Explanation: There is no common prefix among the input strings.
 

Constraints:

1 <= strs.length <= 200
0 <= strs[i].length <= 200
strs[i] cons

````python

def longestCommonPrefix(strs: list[str]) -> str:
    # pick the first word as the prefix as we will use 
    # it as the initial variable to check with
    prefix = strs[0]
    # Start the loop from item[1] as [0] is the prefix so ignore that
    for word in strs[1:]:
        # shorten the prefix until it matches the prefix
        # Else we know there is no matching
        while not word.startswith(prefix):
            prefix = prefix[:-1]
            
            # exit out of the while loop if prefix is depleted
            if not prefix:
                return ""
    return prefix
        

input = ["flower","flow","flight"]
print(longestCommonPrefix(input))
````
