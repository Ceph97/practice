### Missing Number

Write a function to find the missing number in a given integer array of 1 to 100.

The function takes to parameter the array and the number of elements that needs to be in array.  

For example if we want to find missing number from 1 to 6 the second parameter will be 6.

**Example**

```missing_number([1, 2, 3, 4, 6], 6) # 5```

````python
# For this you will need to sum the numbers
# This will be an incremental array since its 1 to 100 therefore it is sorted

def missing_number(arr, number):
    # Sum a full array
    # Then subtract the sum of arr passed
    # The missing number will be the difference
    # On the list generator we add + 1 as range start at 0 to n - 1
    # return sum([i for i in range(number + 1)]) - sum(arr)
    return sum(range(1,number+1)) - sum(arr)

missing_number([1,2,3,4,5,6,8], 8)

````
