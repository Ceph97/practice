## Sum of 2 numbers

Write a program to find all pairs of integers whose sum is equal to a given number.

**EXAMPLE**
```[2, 6, 3, 9, 11]``` 9 → ```[6,3]```

````python
# SUM  OF 2 NUMBERS

def two_sum(array, target):
    tracker = {}

    # Go through all the elements in the array 
    # We need to store them in a dict for easy checks
    # We are enumarating since we need key value pair
    # The index is only needed to complete the pair in tracker dict {value:index}
    # For each iteration we need to find the number we need to get the target
    # current - target = the other number
    for index, value in enumerate(array):
        seen = target - value
        if seen in tracker:
            # return the index or current number and one in tracker
            return [tracker.get(seen), index]
        tracker[value] = index
    return []

two_sum([2,6,3,9,11], 9)
````
