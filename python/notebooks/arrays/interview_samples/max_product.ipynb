## Max Product od Two Integers

Find the maximum product of two integers in an array where all elements are positive.

**Example**
arr = ```[1, 7, 3, 4, 9, 5]``` 
```max_product(arr)``` # Output: 63 (9*7)

````python
# Max product of two integger

# Try with a moving window algorithm
def max_product(arr):
    
    # left pointer
    left = 0

    l_high = 0
    r_high = 0
 

    for right in range(1,len(arr)):
        if left < right:
            curr_left, curr_right = arr[left], arr[right]

            if curr_left > l_high:
                # Only update the value if its 0 if not we know they are duplicate
                # So do nothing
                if curr_left == curr_right and curr_left != 0:
                    l_high = curr_left

            if curr_right > r_high:
                # Avoid non-zero duplicates
                if curr_right == curr_left and curr_right != 0:
                    r_high = curr_right

            left += 1

    return r_high * l_high


max_product([1, 7, 3, 4, 9, 5] )


````

````python
# Hint: Track Highest and second highest number
def max_product(arr):
    highest = 0
    second = 0

    for curr in arr:
        if curr > highest:
            # demote the highest value to be the second highest value
            second = highest
            # then the new highest will be the curr value
            highest = curr
        
        # An edge case like this 9, 1, 7, 3]
        # Where the highest value is caught in the begining of the iteration
        elif curr > second:
            second = curr
    return highest * second
            
max_product([9, 1, 7, 3])
````
