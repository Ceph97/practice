
## Left Rotation



A left rotation operation on an array shifts each of the array's elements 
unit to the left. For example, if  left rotations are performed on array , then the array would become . 
Note that the lowest index item moves to the highest index in a rotation. This is called a circular array.


Given an array  of  integers and a number, , perform  left rotations on the array. Return the updated array to 
be printed as a single line of space-separated integers.

**Example**
```[1,2,3,4,5]``` -> ```[3,4,5,1,2]``` number of rotations ```2```

````python
def left_rotation(arr:list, rotations):
    result = []
    
    for n in range(len(arr)-1, rotations-1, -1 ):
        result.append(arr.pop(n))
        
    result = result[::-1]
    result.extend(arr)
    return result
    

print(left_rotation([1,2,3,4,5],2))
````
