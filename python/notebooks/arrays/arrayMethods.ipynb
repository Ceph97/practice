```python
# Create and trarverse Array
from array import *

# Type codes:
# "i" -> int
# "s" -> string etc 
my_array = array('i', [1,2,3,4,5])

# trave
# for i in my_array:
    # print(i)

# append()
print("======================================")
my_array.append(6)
print(f"Append 11: {my_array}")

print("======================================")
# Extend()
my_array.extend([7,8,9])
print(f"Extend: {my_array}")

print("======================================")
# fromlist()
# Does the same job as the extend.
my_array.fromlist([10,11,12])
print(f"fromlist(): {my_array}")

print("======================================")
# remove()
# Remove first value it finds if duplicates exist
# Doesnt return the value, returns None
my_array.remove(12)
print(f"remove(12): {my_array}")

print("======================================")
# pop()
# Remove last element and returns it
print(f"remove: {my_array.pop()}")

print("======================================")
# index()
# Return index of the value
print(f"Index(0): {my_array.index(10)}")

print("======================================")
# reverse()
# Reverseses the array
my_array.reverse()
print(f"reverse array: {my_array}")

print("======================================")
# buffer_info() method
# Gets array buffer information
print(f"buffer info: {my_array.buffer_info()}")

print("======================================")
# count()
# Count elements occurence
print(f"Count(1): {my_array.count(1)}")

print("======================================")
# Convert Array to Python list 
strTemp = my_array.tolist()
print(f"tolist(): {strTemp}")


print("======================================")
# Slicing elemments
print(f"Slicing elements: {my_array[len(my_array):1:-1]}")

```
