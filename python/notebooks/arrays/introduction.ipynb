# Arrays

- Use numpy to instead of a List for some specific use cases as they its more efficient
- Contigous location in memory; ```[1,2,3,4]```
- Store same data type
- Sequential increase


### WHY WE NEED ARRAYS
- Store collections of variables of the same type instead of storing variables separately.

### TYPES
1. One dimensional
    - ```a[i]``` -> ```i``` between ```0``` and ```n```
    - Exist of one row

2. Multi dimennsional
    - **2 dimensional**: Valued are declared with double index ```a[i][j]```
     - ```i``` is the row number and ```j``` is the column number
     - locate row then column index

    - ***3 dimensional**: Values are declared with triple index ```a[i][j][k]```
     - You can think of it as a cube.
     - Accessing elemnts: 
        - First index is depth.
        - Then Row.
        - Then Column.


````python
# Using Array module
# This module only support basic datatypes and its homoginious

import array

my_array = array.array('i')
print(my_array)

my_array1 = array.array('i', [1,2,3,4])
print(my_array1)
````

````python
# Using numpy
# Its feature rich and effecient
# O(n) space and time depending on the amount of items.
import numpy as np 
np_array = np.array([], dtype=int)
print(np_array)
np_array1 = np.array([1,2,3,4])
print(np_array1)
````

````python
# INSERTING ELEMENTS:
import array

# Insert at 0 in non empty we need to shift items
my_array = array.array('i', [1,2,3,4,5])
my_array.insert(0,6)
print(my_array)

# Insert in the middle only half of items O(n) time and space
# Insert at the end is O(1) not shifting
````

````python
# TRAVERSING AN ARRAY
# time and space is O(n)
import array

# Insert at 0 in non empty we need to shift items
my_array = array.array('i', [1,2,3,4,5])

def traverseArray(array):
    for i in array:
        print(i)

traverseArray(my_array)
````

````python
# ACCESSING CELLS IN ARRAY
# use array[index]
import array

# Insert at 0 in non empty we need to shift items
my_array = array.array('i', [1,2,3,4,5,6])
print(my_array[0])

# Time: O(1)
# Space: O(1)
def accessElement(array, index):
    if index > len(array):
        print('Index OutOfBounds')
    else:
        print(array[index])
        

accessElement(my_array, 5)
````

````python
# SEARCHING AN ELEMENT
# Use linear search
# Compare with value with the current index we are traversing

import array

# Insert at 0 in non empty we need to shift items
my_array = array.array('i', [1,2,3,4,5,6])

# This returns the index of the value that matches our linear search
# Time: O(1)
# Space: O(n)
def linearSearch(arr, target):
    # len() is constant O(1)
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    # Use -1 as a falsy value
    return -1


print(linearSearch(my_array, 10))
````

````python
# Deleting Element of an Array:

# Deleting will shift elements since arrays are contigous like a train
# Its efficient only when you remove last element

import array

# Time: O(n) and last element is O(1)
# Space: O(1)

# Insert at 0 in non empty we need to shift items
my_array = array.array('i', [1,2,3,4,5])

my_array.remove(1)

print(my_array)
````
