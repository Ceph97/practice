# Arrays

- Use numpy to instead of a List for some specific use cases as they its more efficient
- Contigous location in memory; ```[1,2,3,4]```
- Store same data type
- Sequential increase


### WHY WE NEED ARRAYS
- Store collections of variables of the same type instead of storing variables separately.

### TYPES
1. One dimensional
    - ```a[i]``` -> ```i``` between ```0``` and ```n```
    - Exist of one row

2. Multi dimennsional
    - **2 dimensional**: Valued are declared with double index ```a[i][j]```
     - ```i``` is the row number and ```j``` is the column number
     - locate row then column index

    - ***3 dimensional**: Values are declared with triple index ```a[i][j][k]```
     - You can think of it as a cube.
     - Accessing elemnts: 
        - First index is depth.
        - Then Row.
        - Then Column.
