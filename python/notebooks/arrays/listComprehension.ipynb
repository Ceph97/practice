## List Comprehension

- It is a process of creating a new list from an existing one.
- It works on all iterable data types

```python
# Without conditions
new_list = [new_item for item in list]

# With conditions
# The condition is applied on elements as seen on original list.
new_list = [new_item for item in list if condition]
```



````python
prev_list = [1 , 2, 3 , 4]
new_list = [item * 2 for item in prev_list]
print(new_list)
````

````python
prev_list = [1 , 2, 3 , 4]
new_list = [item * 2 for item in prev_list if item % 2 == 0 and item >= 7]
print(new_list)
````

````python
# You can use function as a condition

sentence = "My name is Elshad"

def is_consonant(letter):
    vowels = 'aeiou'
    return letter.isalpha() and letter.lower() not in vowels

consonants = [char for char in sentence if is_consonant(char)]

print(consonants)
````

````python
from random import shuffle
# Having condition before the for loop of the expression checks the conditions against new list

prev_list = [-1, 10, -20, 2, -90, 60, 45, 20]
new_list = [number if number > 0 else "negative number" for number in prev_list]
print(new_list)


#  You can use a function before for loop in the expression
def get_number(number):
    if number > 0:
        return number
    else:
        return "negative number"

new_list = [get_number(number) for number in prev_list]
print(new_list)

shuffle()
````
