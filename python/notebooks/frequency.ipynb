```python
from collections import defaultdict


class Frequency:
    def __init__(self, words: list):
        self.words = words
        self.frequency = {}
        self.tracker = defaultdict(list)
    def top_k(self, k: int) -> list:
        # Step 1: Count total frequency for each word
        for word in self.words:
            self.frequency[word] = self.frequency.get(word, 0) + 1
        # Step 2: Bucket words by their final frequency
        for word, count in self.frequency.items():
            self.tracker[count].append(word)

        # # Step 3: Gather the top k words starting from highest possible frequency
        result = []
        # Why starting in reverse, like the max length of the words?
        # The maximum number of times any single word could possibly appear is len(self.words) (the case where every item in the list is the exact same word).
        # So range(len(self.words), 0, -1) checks frequencies starting from the absolute maximum possible count down to 1.

        for count in range(len(self.words), 0, -1):
            if count in self.tracker:
                for word in self.tracker[count]:
                    result.append(word)
                    # This is used to Kill the loop as the top(K) frequent numbers has been reached.
                    if len(result) == k:
                        return result
                        
        return result
words = [
    "go",
    "python",
    "go",
    "java",
    "go",
    "python"
]
freq = Frequency(words)
print(freq.top_k(2))  # Top 2 most frequent words
# Output: ['go', 'python']
```
