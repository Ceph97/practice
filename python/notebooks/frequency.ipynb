```python
from collections import defaultdict


class Frequency:
    def __init__(self, words: list):
        self.words = words
        self.frequency = {}
        self.tracker = defaultdict(list)
    def top_k(self, k: int) -> list:
        # Step 1: Count total frequency for each word
        for word in self.words:
            self.frequency[word] = self.frequency.get(word, 0) + 1
        # Step 2: Bucket words by their final frequency
        # Frequency should be key and the words will be bucketed
        for word, count in self.frequency.items():
            self.tracker[count].append(word)

        # # Step 3: Gather the top k words starting from highest possible frequency
        result = []
        # Why starting in reverse, like the max length of the words?
        # The maximum number of times any single word could possibly appear is len(self.words) (the case where every item in the list is the exact same word).
        # So range(len(self.words), 0, -1) checks frequencies starting from the absolute maximum possible count down to 1.
        # If we want to find the least. we can start at 0 incrementing the values.
        for count in range(len(self.words), 0, -1):
            if count in self.tracker:
                for word in self.tracker[count]:
                    result.append(word)
                    # This is used to Kill the loop as the top(K) frequent numbers has been reached.
                    if len(result) == k:
                        return result
                        
        return result
words = [
    "go",
    "python",
    "go",
    "java",
    "go",
    "python"
]
freq = Frequency(words)
print(freq.top_k(2))  # Top 2 most frequent words
# Output: ['go', 'python']
```

```python
from collections import defaultdict
class Freq:
    def __init__(self, words:list):
        self.words = words
    def top_K(self, k:int):
        frequency = {}
        # Limits the dict to only store list
        tracker = defaultdict(list)

        # Result of word frequency
        result = []

        # Group words to frequency
        # frequency are like {"word":1, "word2":2}
        for word in self.words:
            frequency[word] = frequency.get(word, 0) + 1
        
        print(frequency)
        
        # Reverse the values to be the keys
        # Bucket it since the renge is known
        # {2:["word1","word2"]}
        for key, value in frequency.items():
            tracker[value].append(key)
        print(tracker)

        # Then return top x words
        for i in range(len(self.words), 0, -1):
            if i in tracker:
                # Get the word in the tracker list
                # tracker element are like {2:["word1","word2"]} etc
                for word in tracker[i]:
                    result.append(word)
                    # Result can only hold k amount of items
                    if len(result) == k:
                        return result
        return result



        
word = [
    "go",
    "python",
    "go",
    "java",
    "go",
    "python",
    "python",
    "python"
]
freq = Freq(word)
print(freq.top_K(1))
```
