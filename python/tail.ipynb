```python
from collections import deque
class TailHeader:
    def __init__(self, filename):
        self.filename = filename
        self.lines = []

    """Reds first n lines of a file."""
    def head(self, n=10) -> list:
        with open(self.filename, "r") as f:
            for line in f.readlines():
                self.lines.append(line)
                if len(self.lines) >= n:
                    break
        return self.lines

    """Reads last n lines of a file"""
    def tail(self, n=10) -> list:
        with open(self.filename, "r") as f:
            return list(deque(f, n))
```

```python
name = TailHeader("tail.txt")
for i in name.head(1):
    print(i.strip())
print("...")
for i in name.tail(10):
    print(i.strip())    
```
